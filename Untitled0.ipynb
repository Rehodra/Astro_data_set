```python
import pandas as pd

# Read your Sector 56 Excel file
df = pd.read_csv("s0056.csv")

print(df.head())

# Take first 100 stars
stars = df.head(100)

print(stars)
```

```python
from astroquery.mast import Catalogs
from astropy.coordinates import SkyCoord
import astropy.units as u

results = []

for _, row in stars.iterrows():

    coord = SkyCoord(
        ra=row["RA"] * u.deg,
        dec=row["DEC"] * u.deg
    )

    neighbors = Catalogs.query_region(
        coord,
        radius=30*u.arcsec,
        catalog="TIC"
    )

    # Remove the target star itself
    if len(neighbors) > 1:

        neighbors = neighbors[1:]

        nearest = min(neighbors["dstArcSec"])

        nearest_mag = neighbors["Tmag"][
            neighbors["dstArcSec"].argmin()
        ]

        count = len(neighbors)

    else:

        nearest = None
        nearest_mag = None
        count = 0

    results.append({
        "TIC_ID": row["#TIC_ID"],
        "Neighbor_Count": count,
        "Nearest_Distance_arcsec": nearest,
        "Nearest_Magnitude": nearest_mag
    })
```

```python
from astroquery.mast import Catalogs
import numpy as np

results = []

for _, row in stars.iterrows():

    coord = SkyCoord(
        ra=row["RA"] * u.deg,
        dec=row["DEC"] * u.deg
    )

    # Query all stars within 30 arcsec
    neighbors = Catalogs.query_region(
        coord,
        radius=30 * u.arcsec,
        catalog="TIC"
    )

    if len(neighbors) == 0:
        continue

    # -------------------------------
    # Identify the Target Star
    # -------------------------------
    target = neighbors[0]

    target_mag = target["Tmag"]
    target_temp = target["Teff"]
    target_radius = target["rad"]
    target_mass = target["mass"]

    # Luminosity is not directly available in TIC
    target_luminosity = np.nan

    # -------------------------------
    # Remove Target Star
    # -------------------------------
    if len(neighbors) > 1:
        neighbor_tbl = neighbors[1:]

        distances = np.array(neighbor_tbl["dstArcSec"])
        magnitudes = np.array(neighbor_tbl["Tmag"])

        neighbor_count = len(neighbor_tbl)

        nearest_idx = np.argmin(distances)

        nearest_distance = distances[nearest_idx]
        nearest_magnitude = magnitudes[nearest_idx]

        brightest_neighbor = np.nanmin(magnitudes)
        mean_neighbor_mag = np.nanmean(magnitudes)

        # Flux = 10^(-0.4 * magnitude)
        fluxes = 10 ** (-0.4 * magnitudes)
        total_neighbor_flux = np.nansum(fluxes)

    else:
        neighbor_count = 0
        nearest_distance = np.nan
        nearest_magnitude = np.nan
        brightest_neighbor = np.nan
        mean_neighbor_mag = np.nan
        total_neighbor_flux = np.nan

    # -------------------------------
    # Store Everything
    # -------------------------------
    results.append({

        # Original Data
        "TIC_ID": row["#TIC_ID"],
        "RA": row["RA"],
        "DEC": row["DEC"],

        # Target Star
        "Target_Magnitude": target_mag,
        "Target_Temperature": target_temp,
        "Target_Radius": target_radius,
        "Target_Mass": target_mass,
        "Target_Luminosity": target_luminosity,

        # Neighbor Statistics
        "Neighbor_Count": neighbor_count,
        "Nearest_Distance_arcsec": nearest_distance,
        "Nearest_Magnitude": nearest_magnitude,
        "Brightest_Neighbor_Magnitude": brightest_neighbor,
        "Mean_Neighbor_Magnitude": mean_neighbor_mag,
        "Total_Neighbor_Flux": total_neighbor_flux,

        # Placeholder Columns (filled later)
        "Sector": 56,
        "LightCurve_File": None,
        "Transit_Detected": None,
        "Detected_Period_days": np.nan,
        "Transit_Depth": np.nan,
        "Transit_Duration_hours": np.nan,
        "Signal_to_Noise": np.nan,
        "Predicted_Class": None,
        "Confidence": np.nan,
        "True_Label": None
    })

# Create DataFrame
final_df = pd.DataFrame(results)

# Save
final_df.to_csv("Sector56_BlendGraph_Dataset.csv", index=False)

print(final_df.head())
```
